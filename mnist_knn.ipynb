### Import necessery Libreries

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

### Read the csv file and drop the labels

```python
train_data = pd.read_csv('mnist_train.csv')
test_data = pd.read_csv('mnist_test.csv')
train_df = pd.DataFrame(train_data) 
test_df = pd.DataFrame(test_data)
train_data.head()
```

```python
train_data_feature = train_df.drop('label', axis=1)
train_data_label = train_df['label']
test_data_feature = test_df.drop('label', axis=1)
test_data_label = test_df['label']
```

### Train the knn model with 3 neighhbors

```python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(train_data_feature,train_data_label)
```

### Test the model and calculate the accuracy

```python
pred_label = model.predict(test_data_feature)
accuracy_score(test_data_label,pred_label)
```
